```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt #импорт библиотек
```

```python
# загрузка датасета
df = pd.read_csv('train.csv')
df
```

```python
# вывод размера таблицы(кол-во строк и столбцов)
df.shape
```

```python
# вывод названий столбцов
df.columns
```

```python
# вывод типов данных столбцов
df.dtypes
```

```python
# вывод кол-ва пассажиров каждого класса
df['Pclass'].value_counts().sort_index()
df
```

```python
# вывод кол-ва мужчин и женщин из датасета
df['Sex'].value_counts()
df
```

```python
# создаем возрастные группы, сделав отдельный датасет, а после распределим данные по этим возрастным группам
bins = [0, 12, 18, 30, 45, 60, 100]
labels = ['0-12', '13-18', '19-30', '31-45', '46-60', '60+']
df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels, right=True)
```

```python
# распределение по возрастным группам
df['AgeGroup'].value_counts().sort_index()
```

```python
# расчет и вывод среднего возраста пассажиров
df['Age'].mean()
```

```python
# расчет и вывод медианного(?) возраста пассажиров
df['Age'].median()
```

```python
# расчет и вывод средней стоимости билета
df['Fare'].mean()
```

```python
# ср. стоимость билетов по классам
fare_by_class = df.groupby('Pclass')['Fare'].mean()
fare_by_class
```

```python
# средний возраст по классам
age_by_class = df.groupby('Pclass')['Age'].mean()
age_by_class
```

```python
# количество мужчин и женщин в каждом классе
sex_by_class = pd.crosstab(df['Pclass'], df['Sex'])
sex_by_class
```

```python
# количество пассажиров по классам
class_counts = df['Pclass'].value_counts().sort_index()
class_counts
```

```python
# класс с наибольшим числом пассажиров
class_counts.idxmax(), "—", class_counts.max(), "пассажиров"
```

```python
# самый дорогой билет
max_fare = df['Fare'].max() #новый столбец по максимальному значению Fare
df.loc[df['Fare'] == max_fare, ['PassengerId', 'Pclass', 'Name', 'Sex', 'Age', 'Fare', 'Survived']]
```

```python
# общее количество выживших и погибших
surv_counts = df['Survived'].value_counts().rename({0: 'Погибшие', 1: 'Выжившие'}) # переменные вместо True/False - 1/0.
surv_counts
```

```python
# выживаемость по полу
surv_sex = df.groupby('Sex')['Survived'].agg(['mean', 'sum', 'count']) # группировка по полу, аргументируется в датафрейме срзнач, суммарное кол-во чел и...подсчет
surv_sex
```

```python
# выживаемость мужчин и женщин по каждому классу
surv_class_sex = df.groupby(['Pclass', 'Sex'])['Survived'].mean().unstack()
surv_class_sex
```

```python
# группа с наибольшей долей выживших
best_group = df.groupby(['Pclass', 'Sex'])['Survived'].mean().idxmax() #лучшая группа по полу среди выживших
best_value = df.groupby(['Pclass', 'Sex'])['Survived'].mean().max() #результат в десятичном значении

best_group, "—", best_value
```

```python
# график количества пассажиров по классам

class_counts.plot(kind='bar') # столбчатая диаграмма по class_count
plt.title('Количество пассажиров по классам') #заголовок
plt.xlabel('Класс') # х - заголовок
plt.ylabel('Количество пассажиров') # у - заголовок
plt.xticks(rotation=0) # подписи по оси х горизонтальные
plt.show()
```

Из гистограммы можно сделать вывод, что все купили дешевые билеты. Почти 500 человек купили билет 3 класса - самого дешевого. более 200 человек - первого класса, а менее 200 - 2 класса.



```python
# график выживаемости по полу

surv_sex['mean'].plot(kind='bar') #столбчатая диаграмма по столбцу mean
plt.title('Доля выживших по полу') #заголовок
plt.xlabel('Пол') # х - заголовок
plt.ylabel('Доля выживших') # у - заголовок
plt.xticks(rotation=0) # подписи по оси х горизонтальные
plt.show()
```

Выживаемость мужчин 20%, женщин - 70%



# Вопрос: влияет ли размер семьи на выживаемость?

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1  #создаем новый столбец FamilySize, где SibSp - кол-во братьев, сестер и супругов на корабле, Parch - кол-во детей/родителей на корабле, +1 - сам пассажир

family_surv = df.groupby('FamilySize')['Survived'].agg(['mean', 'count']) # группировка пассажиров по размеру семьи, берем столбец Survived, где 1- живой, 0 - 200. agg считает срзнач Survived и сколько пассажиров было в группе
print(family_surv)

family_surv['mean'].plot(kind='bar') #столбчатая диаграмма по столбцу mean
plt.title('Доля выживших в зависимости от размера семьи') #заголовок
plt.xlabel('Размер семьи') # х - заголовок
plt.ylabel('Доля выживших') # у - заголовок
plt.xticks(rotation=0) # подписи по оси х горизонтальные
plt.show()
```

Размер семьи влияет на выживаемость в семье
