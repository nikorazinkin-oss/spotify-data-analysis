Датасет отсортирован по мировым продажам в порядке убывания. Выведены 10 игр с максимальными мировыми продажами и только необходимые столбцы.

```python
platform_count.head(10)
```

Было подсчитано количество игр для каждой платформы. Затем платформы были отсортированы по количеству игр, и выведены 10 платформ с наибольшим количеством игр.

```python
platform_sales = df_clean.groupby('Platform')['Global_Sales'].sum()

platform_sales = platform_sales.sort_values(ascending=False)

platform_sales.head(10)
```

Для каждой платформы рассчитана сумма мировых продаж всех игр. После сортировки выведены 10 платформ с наибольшими суммарными продажами.

```python
plt.figure(figsize=(10, 6))

genre_sales.sort_values(ascending=True).plot(kind='barh')

plt.title('Мировые продажи видеоигр по жанрам')
plt.xlabel('Мировые продажи, млн')
plt.ylabel('Жанр')

plt.show()
```

Построена горизонтальная столбчатая диаграмма мировых продаж видеоигр по жанрам. Жанры расположены по величине суммарных продаж.
Жанры с наибольшими продажами: Наибольшие мировые продажи имеют жанры Action, Sports и Shooter. Лидирует жанр Action.

Жанры с наименьшими продажами: Наименьшие продажи имеют жанры Strategy, Adventure и Puzzle. Минимальные суммарные продажи наблюдаются у жанра Strategy.

Разница между значениями: Продажи значительно отличаются между жанрами. У Action продажи составляют примерно 1,7 млрд, а у Strategy — около 0,18 млрд, поэтому разница составляет примерно 1,5 млрд.

```python
correlation = df_clean[['NA_Sales', 'Global_Sales']].corr()

correlation
```

Рассчитан коэффициент корреляции между продажами в Северной Америке и мировыми продажами. Значение коэффициента находится в диапазоне от −1 до +1.

Ответы:

Коэффициент корреляции получился положительным и равен примерно 0,9413.
Да, коэффициент близок к 1.
Да, между NA_Sales и Global_Sales наблюдается выраженная положительная взаимосвязь. Это означает, что игры с большими продажами в Северной Америке, как правило, имеют и большие мировые продажи.

Построена диаграмма рассеяния для визуального исследования взаимосвязи между продажами в Северной Америке и мировыми продажами. По оси X расположены продажи в Северной Америке, по оси Y — мировые продажи.

3. Соответствует ли график результату корреляционного анализа?
Да, график соответствует результату корреляционного анализа. Коэффициент корреляции равен примерно 0,9413, что говорит о сильной положительной взаимосвязи между NA_Sales и Global_Sales.

```python
regional_sales = df_clean[
    [
        'NA_Sales',
        'EU_Sales',
        'JP_Sales',
        'Other_Sales'
    ]
]

regional_sales.corr()
```

Рассчитана корреляционная матрица между продажами в Северной Америке, Европе, Японии и остальных регионах. Таблица позволяет сравнить степень взаимосвязи между региональными продажами.

```python
corr_matrix = regional_sales.corr()

corr_pairs = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

corr_pairs.stack().sort_values(ascending=False)
```

Рассчитана корреляционная матрица между продажами в Северной Америке, Европе, Японии и остальных регионах. Наиболее высокий коэффициент корреляции получен для NA_Sales и EU_Sales.
